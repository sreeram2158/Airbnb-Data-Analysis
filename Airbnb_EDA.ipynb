```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

**Handling missing values**

```python
df['last review']=pd.to_datetime(df['last review'],errors='coerce')
```

```python
df.info()
```

```python
df.fillna({'reviews per month': 0,'last review' : df['last review'].min()},inplace=True)
```

```python
df.dropna(subset=['NAME','host name'],inplace=True)
```

```python
print(df.isnull().sum())
```

```python
df=df.drop(columns=["license","house_rules"],errors='ignore')
```

```python
df.head()
```

```python
# remove dollar signs and convert to float

df['price']=df['price'].replace('[\$,]','',regex=True).astype(float)
df['service fee']=df['service fee'].replace('[\$,]','',regex=True).astype(float)
```

```python
df.head()
```

**remove duplicates**

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

**descriptive statistics**

```python
df.describe()
```

**Visualization**

**What is the distibution of listing prices?**

```python
plt.figure(figsize=(14,8))
sns.histplot(df['price'],bins=50,kde=True)
plt.title("Distribution of listing prices")
plt.xlabel("Price $")
plt.ylabel("Frequency")
plt.show()
```

**How are different room types distributed**

```python
df['room type']
```

```python
plt.figure(figsize=(14,8))
sns.countplot(x="room type",data=df)
plt.title("Room type distribution")
plt.xlabel("Room type")
plt.ylabel("Count")
plt.show()
```

**How are listings distributed across different neighbourhoods?**

```python
plt.figure(figsize=(14,8))
sns.countplot(y="neighbourhood group",data=df,order=df['neighbourhood group'].value_counts().index)
plt.title('Number of listings by Neighbourhood Group')
plt.xlabel('count')
plt.ylabel('Neighbourhood Group')
plt.show()
```

**What is the relationship btw price and room type**

```python
plt.figure(figsize=(14,8))
sns.boxplot(x='room type',y='price',data=df,hue='room type')
plt.title('Price vs Room Type')
plt.xlabel('Room Type')
plt.ylabel('Price($)')
plt.legend(title='Room Type')
plt.show()
```

**How has the number of reviews change over the time?**

```python
df['last review']=pd.to_datetime(df['last review'])
reviews_over_time=df.groupby(df['last review'].dt.to_period('M')).size()

plt.figure(figsize=(12,6))
reviews_over_time.plot(kind='line')
plt.title('Number of Reviews Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Reviews')
plt.show()
```
